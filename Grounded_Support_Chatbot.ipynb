```python
!pip install -q google-genai

from google import genai
from google.colab import userdata

api_key = userdata.get("GEMINI_API_KEY")

client = genai.Client(api_key=api_key)

knowledge_base = """
Q: What courses are available at NextStep NEET & JEE Academy?
A: We offer NEET preparation, JEE Main preparation, JEE Advanced preparation, and foundation courses for classes 11 and 12.

Q: What are the timings of the academy?
A: The academy is open from 8:00 AM to 8:00 PM from Monday to Saturday.

Q: Is the academy open on Sunday?
A: No, regular classes are not conducted on Sunday.

Q: What is the fee for the NEET course?
A: The NEET one-year classroom course fee is ₹45,000.

Q: What is the fee for the JEE Main course?
A: The JEE Main one-year classroom course fee is ₹50,000.

Q: What is the fee for the JEE Advanced course?
A: The JEE Advanced one-year classroom course fee is ₹60,000.

Q: Can the course fee be paid in installments?
A: Yes, students can pay the course fee in 3 installments.

Q: How long is the one-year course?
A: The one-year course runs for 12 months.

Q: How many classes are conducted in a week?
A: Regular students have classes 6 days a week from Monday to Saturday.

Q: Are study materials provided?
A: Yes, printed study materials and practice sheets are provided to enrolled students.

Q: Are mock tests conducted?
A: Yes, full-length mock tests are conducted twice every month.

Q: How can I join the academy?
A: You can visit the academy office, complete the admission form, submit the required documents, and pay the first installment.

Q: What documents are required for admission?
A: Students need a passport-size photo, a copy of their previous mark sheet, and a valid ID proof.

Q: Is there a scholarship test?
A: Yes, the academy conducts a scholarship test once every 3 months.

Q: What is the maximum scholarship available?
A: Students can receive up to 50% scholarship based on their scholarship test performance.

Q: Are doubt-clearing sessions available?
A: Yes, doubt-clearing sessions are conducted every Wednesday and Saturday.

Q: How long is each regular class?
A: Each regular class is usually 2 hours long.

Q: Are classes available for class 11 students?
A: Yes, foundation and integrated preparation courses are available for class 11 students.

Q: Are classes available for class 12 students?
A: Yes, NEET and JEE preparation courses are available for class 12 students.

Q: Can students attend a demo class?
A: Yes, students can attend 1 demo class before completing admission.

Q: Is attendance compulsory?
A: Students are expected to maintain at least 75% attendance.

Q: Can I get a refund after joining the course?
A: Fee refunds are not available after 7 days from the date of admission.

Q: How can I contact the academy?
A: You can contact the academy at support@nextstepacademy.com.
"""

system_instruction = f"""
You are the support chatbot for NextStep NEET & JEE Academy.

Rules:
1. Answer only using the information given in the knowledge base.
2. Do not use outside knowledge.
3. Do not make up any information.
4. If the answer is not available in the knowledge base, say exactly:
"I do not have information about that. Please contact us at support@nextstepacademy.com."
5. Be helpful and answer clearly.
6. Remember the previous conversation and answer based on the chat history when relevant.

Knowledge base:

{knowledge_base}
"""

chat = client.chats.create(
   model="gemini-3.6-flash",
    config={
        "system_instruction": system_instruction,
        "temperature": 0.2
    }
)

while True:
    user_input = input("You: ")

    if user_input.lower() == "quit" or user_input.lower() == "exit":
        print("Bot: Thank you for contacting NextStep NEET & JEE Academy.")
        break

    response = chat.send_message(user_input)
    print("Bot:", response.text)
```
